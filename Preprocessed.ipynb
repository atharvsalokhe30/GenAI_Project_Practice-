```python
import os
import cv2
import numpy as np
from matplotlib import pyplot as plt

Path = "/content/drive/MyDrive/GenAI/OriginalImages"
```

```python
input_folder = '/content/drive/MyDrive/GenAI/OriginalImages'
output_folder = '/content/drive/MyDrive/GenAI/PreprocessedImages'
if not os.path.exists(output_folder):
    os.makedirs(output_folder)
```

```python
def preprocess_image(image_path):
    img = cv2.imread(image_path)

    img_resized = cv2.resize(img, (256, 256))

    img_gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

    img_normalized = img_gray / 255.0

    return img_normalized
```

```python
def preprocess_image(image_path):
    img = cv2.imread(image_path)

    img_resized = cv2.resize(img, (256, 256))

    img_gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

    img_normalized = img_gray / 255.0

    return img_normalized
```

```python
for img_name in os.listdir(input_folder):
    img_path = os.path.join(input_folder, img_name)

    if not (img_name.endswith('.jpg') or img_name.endswith('.png') or img_name.endswith('.jpeg')):
        continue

    processed_img = preprocess_image(img_path)

    output_path = os.path.join(output_folder, img_name)
    cv2.imwrite(output_path, (processed_img * 255).astype('uint8'))

print(" All images preprocessed and saved successfully!")
```
